**Stemming**

```python
#stemming
words =["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
#regexpstemmer class - the RegexpStemmer is configured to remove common suffixes like 
#"ing", "s", "e", and "able" from words, based on a regular expression pattern.
# The RegexpStemmer will attempt to stem a word by removing any of the specified suffixes,
# provided the word is at least 4 characters long.

from nltk.stem import RegexpStemmer
stemmings = RegexpStemmer('ing$|s$|e$|able$',min=4)

#'ing$': This matches any word that ends with "ing".
# 's$': This matches any word that ends with "s".
# 'e$': This matches any word that ends with "e".
# 'able$': This matches any word that ends with "able".
```

```python
words = ['running', 'cats', 'able', 'dance', 'playing', 'happiness']
for word in words:
    print(word+'----->'+ stemmings.stem(word))
```

```python
#snowball stemmer
from nltk.stem import SnowballStemmer
snow_ball = SnowballStemmer('english')
words = ['running', 'cats', 'able', 'dance', 'playing', 'happiness']
for word in words:
    print(word+'----->'+ snow_ball.stem(word))
```

```python
stemmings.stem('sportingly'),stemmings.stem('fairly')
```

```python
snow_ball.stem('sportingly'),snow_ball.stem('fairly')
```

```python
#apply stopwords and then filter using stemming

from nltk.stem import SnowballStemmer # best out of the three stemming methods 
from nltk.tokenize import sent_tokenize , word_tokenize
#corpus
paragraph ="Dream, dream, dream. Dreams transform into thoughts and thoughts result in action. If you fail, never give up because F.A.I.L. means 'First Attempt In Learning'. End is not the end, in fact, E.N.D. means 'Effort Never Dies'. If you get no as an answer, remember N.O. means 'Next Opportunity'. God gives us the ability to do, and we do the best we can with our ability. It is the quality of our work that will speak for us. Great dreams of great dreamers are always transcended. A dream is not that which you see while sleeping, it is something that does not let you sleep. The best way to predict your future is to create it. Let us sacrifice our today so that our children can have a better tomorrow. You have to dream before your dreams can come true. If you want to shine like a sun, first burn like a sun. Be more dedicated to making solid achievements than in running after swift but synthetic happiness. Man needs his difficulties because they are necessary to enjoy success. You have to face difficulties in life. Don’t give up. A person who has never made a mistake has never tried anything new."

sentences = sent_tokenize(paragraph) #list of sentences tokenized from the paragrapgh
stemmer= SnowballStemmer('english')

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])#tokenization Sentences-->words
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]= ' '.join(words)# converting all the list of words into senetnces
print(sentences)
```

```python
#apply sstopwords and then filter using lemmatization
#using lemmatization which is better than stemming

from nltk.stem import WordNetLemmatizer 
from nltk.tokenize import sent_tokenize , word_tokenize
#corpus
paragraph ="Dream, dream, dream. Dreams transform into thoughts and thoughts result in action. If you fail, never give up because F.A.I.L. means 'First Attempt In Learning'. End is not the end, in fact, E.N.D. means 'Effort Never Dies'. If you get no as an answer, remember N.O. means 'Next Opportunity'. God gives us the ability to do, and we do the best we can with our ability. It is the quality of our work that will speak for us. Great dreams of great dreamers are always transcended. A dream is not that which you see while sleeping, it is something that does not let you sleep. The best way to predict your future is to create it. Let us sacrifice our today so that our children can have a better tomorrow. You have to dream before your dreams can come true. If you want to shine like a sun, first burn like a sun. Be more dedicated to making solid achievements than in running after swift but synthetic happiness. Man needs his difficulties because they are necessary to enjoy success. You have to face difficulties in life. Don’t give up. A person who has never made a mistake has never tried anything new."

sentences = sent_tokenize(paragraph) #list of sentences tokenized from the paragrapgh
lemmatizer= WordNetLemmatizer()

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])#tokenization Sentences-->words
    words = [lemmatizer.lemmatize(word,pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]= ' '.join(words)# converting all the list of words into senetnces
print(sentences)
```


**NAME ENTITITY RECOGNITION**

Named Entity Recognition helps in extracting structured information from unstructured text and is widely used in tasks like information retrieval, question answering, and summarization.

```python
import nltk
nltk.download('maxent_ne_chunker')
nltk.download('words')
```

```python
from nltk import word_tokenize, pos_tag, ne_chunk

sentence = "Barack Obama was born on August 4, 1961, in Honolulu, Hawaii."
words = word_tokenize(sentence)
tagged_tokens = nltk.pos_tag(words)
ner_tree = nltk.ne_chunk(tagged_tokens)
print(ner_tree)
```
